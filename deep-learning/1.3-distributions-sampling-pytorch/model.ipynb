```python
import torch
import torch.distributions as dist
```

```python
# ceate a Bernoulli distribution

bernoulli = dist.Bernoulli(torch.tensor([0.5]))
# print(bernoulli)

# Sample from the distribution
for i in range(10):
    print(bernoulli.sample())
```

```python
samples = [bernoulli.sample() for i in range(1000)]
torch.stack(samples) # stacks tensors on top of each other
```

```python
print(f"Mean: {torch.mean(torch.stack(samples))}")
```

```python
print(f"var: {torch.var(torch.stack(samples))}")
```

```python
# create a Normal distribution
normal = dist.Normal(torch.tensor([0.0]), torch.tensor([1.0])) # (mean, std)

for i in range(10):
    print(normal.sample())
```

```python
samples = [normal.sample() for i in range(1000)]
print(f"Mean: {torch.mean(torch.stack(samples))}")
print(f"var: {torch.var(torch.stack(samples))}")
```

## In sampling any prob distribution, the samples NEVER FULLY MATCH the ORIGINAL Distribution !!!!

So the examples above shows that, where each output is never the same and is very close to the given mean tensor but never the exact same
