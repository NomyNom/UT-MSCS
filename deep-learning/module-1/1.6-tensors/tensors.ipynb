Tensors:

- N-D array

- 1D tensor: Vector
- 2D tensor: Matrix
- 3D tensor: Image
- 4D tensor: Video

A 'torch.Tensor' is a data container

- shape: the size of tensor
- dtype: the data type
- ndim: the number of dimensions

```python
import torch
from PIL import Image
import numpy as np
```

```python
img = Image.open('./data/cat.jpeg')

# convert to tensor
cat = torch.as_tensor(np.array(img)).float()
print(cat.shape)
print(cat.dtype) # uint8 == default for images
print(cat.ndim)
```
